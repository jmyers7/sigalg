The `Event` class in SigAlg represents events, which are subsets of a sample space. The API reference is [here](https://johnmyers-phd.com/sigalg/api/core/#sigalg.core.Event).

## Mathematical definition

Let $\mathcal{F}$ be a $\sigma$-algebra on a sample space $\Omega$. An *event* (relative to $\mathcal{F}$) is a subset $A$ of $\Omega$ in $\mathcal{F}$. In general measure theory, an event is called an $\mathcal{F}$-measurable set.

However, in SigAlg, we do *not* enforce the requirement that an event is measurable with respect to a pre-existing $\sigma$-algebra. Any subset of a sample space can be represented as an instance of `Event`.

## API examples

### Creating events

#### Using the Event constructor

Create an event by calling the `Event` constructor:

```python
from sigalg.core import Event, SampleSpace

Omega = SampleSpace().from_sequence(size=4)

# Create an event directly
A = Event(sample_space=Omega, name="A").from_list([0, 2])
print(A)
```

#### Using `get_event` from a sample space

The more convenient way is to extract an event directly from the sample space:

```python
B = Omega.get_event([1, 3], name="B")
print(B)
```

#### Using indexing notation

You can also use indexing notation on the sample space:

```python
# Using a list of positions
C = Omega[[0, 1], "C"]
print(C, "\n")

# Using a slice
D = Omega[1:3, "D"]
print(D)
```

### Set-theoretic operations

Events support standard set operations like union, intersection, complement, and difference.

#### Union

The union of two events $A$ and $B$ is the set of all outcomes in either $A$ or $B$ (or both):

```python
Omega = SampleSpace().from_sequence(size=5)

A = Omega.get_event([0, 1, 2], name="A")
B = Omega.get_event([2, 3, 4], name="B")

# Using the union method
union_1 = A.union(B)
print(union_1, "\n")

# Using the | operator
union_2 = A | B
print(union_2)
```

#### Intersection

The intersection of two events $A$ and $B$ is the set of all outcomes in both $A$ and $B$:

```python
# Using the intersection method
intersection_1 = A.intersection(B)
print(intersection_1, "\n")

# Using the & operator
intersection_2 = A & B
print(intersection_2)
```

#### Complement

The complement of an event $A$ is the set of all outcomes not in $A$:

```python
A = Omega.get_event([0, 2], name="A")

# Using the complement method
complement_1 = A.complement()
print(complement_1, "\n")

# Using the ~ operator
complement_2 = ~A
print(complement_2)
```

#### Difference

The difference $A \setminus B$ is the set of all outcomes in $A$ but not in $B$:

```python
A = Omega.get_event([0, 1, 2], name="A")
B = Omega.get_event([1, 2, 3], name="B")

# Using the difference method
difference_1 = A.difference(B)
print(difference_1, "\n")

# Using the - operator
difference_2 = A - B
print(difference_2)
```

### Subset and superset relations

Check if one event is a subset or superset of another:

```python
Omega = SampleSpace().from_sequence(size=5)

A = Omega.get_event([1, 2], name="A")
B = Omega.get_event([0, 1, 2, 3], name="B")
C = Omega.get_event([3, 4], name="C")

# Subset checks
print(f"Is A ⊆ B? {A <= B}")
print(f"Is A ⊂ B (proper subset)? {A < B}")

# Superset checks
print(f"Is B ⊇ A? {B >= A}")
print(f"Is B ⊃ A (proper superset)? {B > A}")

# Non-overlapping events
print(f"Is A ⊆ C? {A <= C}")
```

### Properties and methods of events

#### Accessing the sample space

Every event knows its sample space:

```python
Omega = SampleSpace(name="Omega").from_sequence(size=4)
A = Omega.get_event([0, 1], name="A")

print(A.sample_space)
```

#### Converting to a sample space

Convert an event to a sample space containing only the outcomes in the event:

```python
Omega = SampleSpace().from_sequence(size=5)
A = Omega.get_event([1, 3], name="A")

new_space = A.to_sample_space()
print(new_space)
```

### Testing set-theoretic properties

#### De Morgan's laws

De Morgan's laws state that:
- $(A \cup B)^c = A^c \cap B^c$
- $(A \cap B)^c = A^c \cup B^c$

Let's verify these:

```python
Omega = SampleSpace().from_sequence(size=6)

A = Omega.get_event([0, 1, 2], name="A")
B = Omega.get_event([2, 3, 4], name="B")

# First law: (A ∪ B)^c = A^c ∩ B^c
left_1 = ~(A | B)
right_1 = ~A & ~B
print("First De Morgan's law:")
print(f"(A ∪ B)^c = {left_1}")
print(f"A^c ∩ B^c = {right_1}")
print(f"Equal? {left_1 == right_1}\n")

# Second law: (A ∩ B)^c = A^c ∪ B^c
left_2 = ~(A & B)
right_2 = ~A | ~B
print("Second De Morgan's law:")
print(f"(A ∩ B)^c = {left_2}")
print(f"A^c ∪ B^c = {right_2}")
print(f"Equal? {left_2 == right_2}")
```

#### Distributive laws

The distributive laws state that:
- $A \cup (B \cap C) = (A \cup B) \cap (A \cup C)$
- $A \cap (B \cup C) = (A \cap B) \cup (A \cap C)$

```python
Omega = SampleSpace().from_sequence(size=6)

A = Omega.get_event([0, 1], name="A")
B = Omega.get_event([1, 2, 3], name="B")
C = Omega.get_event([3, 4, 5], name="C")

# First law
left = A | (B & C)
right = (A | B) & (A | C)
print(f"A ∪ (B ∩ C) = (A ∪ B) ∩ (A ∪ C)? {left == right}\n")

# Second law
left = A & (B | C)
right = (A & B) | (A & C)
print(f"A ∩ (B ∪ C) = (A ∩ B) ∪ (A ∩ C)? {left == right}")
```
