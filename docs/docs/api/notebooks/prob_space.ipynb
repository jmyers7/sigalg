The `ProbabilitySpace` class in SigAlg represents a probability space, which is the fundamental mathematical structure for probability theory. The API reference is [here](https://johnmyers-phd.com/sigalg/api/core/#sigalg.core.ProbabilitySpace).

## Mathematical definition

A *probability space* is a triple $(\Omega, \mathcal{F}, P)$ consisting of a sample space $\Omega$, a $\sigma$-algebra $\mathcal{F}$ on $\Omega$, and a probability measure $P$ defined on $\mathcal{F}$.

## API examples

### Creating probability spaces

#### With default components

Create a probability space with default uniform probability measure and power-set $\sigma$-algebra:

```python
from sigalg.core import ProbabilitySpace, SampleSpace

Omega = SampleSpace().from_sequence(size=3)

# Default: uniform probability, power-set sigma-algebra
prob_space = ProbabilitySpace(sample_space=Omega)
print(prob_space)
```

#### With custom components

Create a probability space with custom probability measure and $\sigma$-algebra:

```python
from sigalg.core import ProbabilityMeasure, SigmaAlgebra

Omega = SampleSpace().from_sequence(size=3)

# Custom probability measure
P = ProbabilityMeasure(sample_space=Omega).from_dict({0: 0.5, 1: 0.3, 2: 0.2})

# Custom sigma-algebra
F = SigmaAlgebra(sample_space=Omega, name="F").from_dict({0: 0, 1: 1, 2: 1})

prob_space = ProbabilitySpace(
    sample_space=Omega, sigma_algebra=F, probability_measure=P
)
print(prob_space)
```

#### From a dictionary

Create a probability space directly from a dictionary of probabilities:

```python
prob_space = ProbabilitySpace().from_dict(
    {
        "a": 0.5,
        "b": 0.3,
        "c": 0.2,
    }
)
print(prob_space)
```

#### From a sample space

Use the `make_probability_space` method on a sample space:

```python
Omega = SampleSpace().from_list(["red", "green", "blue"])

# With defaults
prob_space = Omega.make_probability_space()
print(prob_space)
```

### Properties of probability spaces

#### Accessing components

Access the sample space, $\sigma$-algebra, and probability measure:

```python
Omega = SampleSpace(name="Omega").from_sequence(size=3)
P = ProbabilityMeasure(sample_space=Omega, name="P").from_dict({0: 0.2, 1: 0.5, 2: 0.3})
F = SigmaAlgebra(sample_space=Omega, name="F").from_dict({0: 0, 1: 1, 2: 1})

prob_space = ProbabilitySpace(
    sample_space=Omega, sigma_algebra=F, probability_measure=P
)

print(prob_space.sample_space, "\n")
print(prob_space.sigma_algebra, "\n")
print(prob_space.probability_measure)
```

#### Modifying components

Update the probability measure or $\sigma$-algebra:

```python
# Create a new probability measure
Q = ProbabilityMeasure(sample_space=Omega, name="Q").from_dict({0: 0.6, 1: 0.3, 2: 0.1})

# Create a new sigma-algebra
G = SigmaAlgebra(sample_space=Omega, name="G").from_dict({0: 1, 1: 0, 2: 1})

# Update the probability space
prob_space.probability_measure = Q
prob_space.sigma_algebra = G

print(prob_space)
```

### Computing probabilities

#### Probability of individual outcomes

Compute probabilities of individual sample points:

```python
prob_space = ProbabilitySpace().from_dict(
    {
        "win": 0.3,
        "lose": 0.5,
        "tie": 0.2,
    }
)

print(f"P(win) = {prob_space.probability_measure('win')}")
print(f"P(lose) = {prob_space.probability_measure('lose')}")
print(f"P(tie) = {prob_space.probability_measure('tie')}")
```

#### Probability of events

Compute probabilities of events:

```python
Omega = SampleSpace().from_sequence(size=6, initial_index=1)
P = ProbabilityMeasure.uniform(sample_space=Omega, name="P")

prob_space = ProbabilitySpace(sample_space=Omega, probability_measure=P)

# Event: rolling an even number
even = prob_space.get_event([2, 4, 6], name="Even")
print(f"P(Even) = {prob_space.probability_measure(even)}\n")

# Event: rolling > 4
high = prob_space.get_event([5, 6], name="High")
print(f"P(High) = {prob_space.probability_measure(high)}\n")

# Event: even OR high
even_or_high = even | high
print(f"P(Even or High) = {prob_space.probability_measure(even_or_high)}")
```

### Conditional probabilities

Compute conditional probabilities using the formula $P(A|B) = \frac{P(A \cap B)}{P(B)}$:

```python
Omega = SampleSpace().from_sequence(size=4)
P = ProbabilityMeasure(sample_space=Omega).from_dict({0: 0.1, 1: 0.2, 2: 0.3, 3: 0.4})

prob_space = ProbabilitySpace(sample_space=Omega, probability_measure=P)

A = prob_space.get_event([0, 1], name="A")
B = prob_space.get_event([1, 2], name="B")

# Compute conditional probability
cond_prob = prob_space.conditional_probability(event=A, given=B)
print(f"P(A|B) = {cond_prob}\n")

# Verify the formula
print(f"P(A ∩ B) = {P(A & B)}")
print(f"P(B) = {P(B)}")
print(f"P(A ∩ B) / P(B) = {P(A & B) / P(B)}")
```

### Independence

#### Independence of events

Check if two events are independent:

```python
from scipy.stats import bernoulli

from sigalg.core import Time
from sigalg.processes import IIDProcess

# Flip a biased coin twice
time = Time.discrete(length=1)
coin_flips = IIDProcess(
    distribution=bernoulli(p=0.7),
    support=[0, 1],
    name="coin_flips",
    time=time,
).from_enumeration()

Omega = coin_flips.domain
P = coin_flips.probability_measure

prob_space = ProbabilitySpace(sample_space=Omega, probability_measure=P)

print(coin_flips, "\n")

# Define events
first_heads = prob_space.get_event([2, 3], name="first_heads")
second_heads = prob_space.get_event([1, 3], name="second_heads")

print(
    f"Are 'first_heads' and 'second_heads' independent? {prob_space.are_independent(event1=first_heads, event2=second_heads)}"
)
```

#### Independence of random variables

Check if two random variables are independent:

```python
# Extract the random variables
flip1, flip2 = coin_flips

print(f"Are flip1 and flip2 independent? {P.are_independent(rv1=flip1, rv2=flip2)}\n")

# Create a dependent random variable
sum_flips = flip1 + flip2
print(
    f"Are flip1 and sum_flips independent? {P.are_independent(rv1=flip1, rv2=sum_flips)}"
)
```

### Sampling from probability spaces

Generate random samples from the probability space:

```python
import numpy as np

Omega = SampleSpace().from_sequence(size=4)
P = ProbabilityMeasure(sample_space=Omega).from_dict(
    {0: 0.15, 1: 0.25, 2: 0.35, 3: 0.25}
)

prob_space = ProbabilitySpace(sample_space=Omega, probability_measure=P)

# Generate samples with a fixed seed for reproducibility
rng = np.random.default_rng(seed=42)
samples = prob_space.sample(size=10, random_state=rng)

print(f"Generated samples: {samples}")
```

Verify that the empirical frequencies approximate the true probabilities:

```python
import pandas as pd

# Generate many samples
rng = np.random.default_rng(seed=42)
large_sample = prob_space.sample(size=1000, random_state=rng)

empirical = pd.Series(large_sample).value_counts(normalize=True).sort_index()

print("Comparison of true vs empirical probabilities:")
for outcome in Omega:
    true_prob = P(outcome)
    emp_prob = empirical.get(outcome, 0)
    print(f"  P({outcome}): true = {true_prob:.3f}, empirical = {emp_prob:.3f}")
```

### Testing probability axioms

#### Axiom 1: Non-negativity

All probabilities are non-negative:

```python
prob_space = ProbabilitySpace().from_dict({0: 0.2, 1: 0.5, 2: 0.3})

all_non_negative = all(
    prob_space.probability_measure(omega) >= 0 for omega in prob_space.sample_space
)
print(f"All probabilities non-negative? {all_non_negative}")
```

#### Axiom 2: Normalization

The probability of the entire sample space is 1:

```python
entire_space = prob_space.get_event(list(prob_space.sample_space))
print(f"P(Omega) = {prob_space.probability_measure(entire_space)}")
```

#### Axiom 3: Additivity

For disjoint events, $P(A \cup B) = P(A) + P(B)$:

```python
prob_space = ProbabilitySpace().from_dict({0: 0.2, 1: 0.3, 2: 0.5})

# Disjoint events
A = prob_space.get_event([0], name="A")
B = prob_space.get_event([1], name="B")

# Check: P(A ∪ B) = P(A) + P(B)
union_prob = prob_space.probability_measure(A | B)
sum_prob = prob_space.probability_measure(A) + prob_space.probability_measure(B)

print(f"P(A ∪ B) = {union_prob}")
print(f"P(A) + P(B) = {sum_prob}")
print(f"Equal? {abs(union_prob - sum_prob) < 1e-10}")
```
