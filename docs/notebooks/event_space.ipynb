The `MeasurableSpace` class in SigAlg represents an event space, which is a sample space equipped with a $\sigma$-algebra. The API reference is [here](https://johnmyers-phd.com/sigalg/api/modules/core/#sigalg.core.MeasurableSpace).

## Mathematical definition

An *event space* is a pair $(\Omega, \mathcal{F})$ consisting of a sample space $\Omega$ and a $\sigma$-algebra $\mathcal{F}$ on $\Omega$. In general measure theory, this is just called a *measurable space*, but the terminology used here is intended to reflect the probabilistic context.

## API examples

### Creating event spaces

#### With default power-set $\sigma$-algebra

Create an event space with the default power-set $\sigma$-algebra (the finest possible $\sigma$-algebra):

```python
from sigalg.core import MeasurableSpace, SampleSpace

Omega = SampleSpace().from_sequence(size=3)

# If no sigma-algebra is provided, a power-set sigma-algebra is created
measurable_space = MeasurableSpace(sample_space=Omega)
print(measurable_space)
```

#### With custom $\sigma$-algebra

Create an event space with a custom $\sigma$-algebra:

```python
from sigalg.core import SigmaAlgebra

Omega = SampleSpace().from_sequence(size=3)

# Create a sigma-algebra with two atoms: {0, 1} and {2}
F = SigmaAlgebra(sample_space=Omega, name="F").from_dict(
    {
        0: 0,
        1: 0,
        2: 1,
    }
)

measurable_space = MeasurableSpace(sample_space=Omega, sig_alg=F)
print(measurable_space)
```

#### From a sample space

Use the `make_measurable_space` method on a sample space:

```python
Omega = SampleSpace().from_list(["s0", "s1", "s2", "s3"])

# Create with default power-set sigma-algebra
measurable_space = Omega.make_measurable_space()
print(measurable_space)
```

### Properties of event spaces

#### Accessing the sample space

Access the underlying sample space:

```python
Omega = SampleSpace(name="Omega").from_sequence(size=4)
F = SigmaAlgebra(sample_space=Omega, name="F").from_dict({0: 0, 1: 0, 2: 1, 3: 1})

measurable_space = MeasurableSpace(sample_space=Omega, sig_alg=F)

print(measurable_space.sample_space)
```

#### Accessing the $\sigma$-algebra

Access the $\sigma$-algebra:

```python
print(measurable_space.sig_alg)
```

#### Modifying the $\sigma$-algebra

You can update the $\sigma$-algebra after creation:

```python
# Create a new, coarser sigma-algebra
G = SigmaAlgebra.trivial(sample_space=Omega, name="G")

# Update the event space
measurable_space.sig_alg = G

print(measurable_space)
```

### Converting to a probability space

#### With uniform probability measure

Convert an event space to a probability space with a uniform distribution:

```python
Omega = SampleSpace().from_sequence(size=3)
F = SigmaAlgebra(sample_space=Omega, name="F").from_dict({0: 0, 1: 0, 2: 1})

measurable_space = MeasurableSpace(sample_space=Omega, sig_alg=F)

# Create probability space with uniform distribution
prob_space = measurable_space.make_probability_space()
print(prob_space)
```

#### With custom probability measure

Convert an event space to a probability space with a custom probability measure:

```python
from sigalg.core import ProbabilityMeasure

# Create custom probability measure
P = ProbabilityMeasure(sample_space=Omega).from_dict({0: 0.2, 1: 0.5, 2: 0.3})

# Create probability space with custom measure
prob_space = measurable_space.make_probability_space(prob_measure=P)
print(prob_space)
```

### Working with events in the event space

#### Creating events

Create events using the event space's sample space:

```python
Omega = SampleSpace().from_sequence(size=5)
measurable_space = MeasurableSpace(sample_space=Omega)

# Create events
A = measurable_space.get_event([0, 1, 2], name="A")
B = measurable_space.get_event([2, 3, 4], name="B")

print(A, "\n")
print(B)
```

#### Checking measurability

We can check if an event is measurable with respect to the $\sigma$-algebra carried by an event space:

```python
Omega = SampleSpace().from_sequence(size=4)

# Create a sigma-algebra with atoms {0, 1} and {2, 3}
F = SigmaAlgebra(sample_space=Omega, name="F").from_dict({0: 0, 1: 0, 2: 1, 3: 1})

measurable_space = MeasurableSpace(sample_space=Omega, sig_alg=F)

# This event is measurable (it's exactly one atom)
A = measurable_space.get_event([0, 1], name="A")
print(A)
print(f"Is A measurable with respect to F? {measurable_space.is_measurable(A)}\n")

# This event is also measurable (it's the union of both atoms)
B = measurable_space.get_event([0, 1, 2, 3], name="B")
print(B)
print(f"Is B measurable with respect to F? {measurable_space.is_measurable(B)}\n")

# Attempting to create a non-measurable event raises ValueError
# (This event would split an atom {0, 1})
try:
    C = measurable_space.get_event([0, 2], name="C")
    print(C)
except ValueError as e:
    print(f"ValueError raised: {e}")
```
