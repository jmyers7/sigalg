```python
import sigalg as sa

Omega = sa.SampleSpace.from_sequence(size=7)
F = sa.SigmaAlgebra(
    domain=Omega,
    mapping={
        0: 0,
        1: 1,
        2: 2,
        3: 2,
        4: 3,  # null atom
        5: 4,  # null atom
        6: 4,  # null atom
    },
)
P = sa.ProbabilityMeasure(
    domain=F,
    mapping={
        0: 0.4,
        1: 0.6,
        2: 0.0,
        3: 0.0,
        4: 0.0,
    },
)
mu = sa.Measure(
    domain=F,
    mapping={
        0: 1,
        1: 2,
        2: 3,
        3: 4,
        4: 0,  # null atom
    },
)
G = sa.SigmaAlgebra(
    domain=Omega,
    mapping={
        0: 0,
        1: 0,
        2: 1,  # null atom
        3: 1,  # null atom
        4: 2,  # null atom
        5: 2,  # null atom
        6: 2,  # null atom
    },
    name="G",
)
```

```python
mu.data.where(mu.data == 0, 1 / (mu.data != 0).sum())
```

```python
print(P.conditional(G, base_measure=mu))
```
