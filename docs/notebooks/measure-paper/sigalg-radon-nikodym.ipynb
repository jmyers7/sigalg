```python
import sigalg as sa

Omega = sa.SampleSpace.from_sequence(size=7)
F = sa.SigmaAlgebra(
    domain=Omega,
    mapping={
        0: 0,
        1: 1,
        2: 2,
        3: 2,
        4: 3,  # null atom
        5: 4,  # null atom
        6: 4,  # null atom
    },
)
P = sa.ProbabilityMeasure(
    domain=F,
    mapping={
        0: 0.2,
        1: 0.6,
        2: 0.2,
        3: 0.0,
        4: 0.0,
    },
)
mu = sa.Measure(
    domain=F,
    mapping={
        0: 1,
        1: 2,
        2: 3,
        3: 4,
        4: 0,  # null atom
    },
)
G = sa.SigmaAlgebra(
    domain=Omega,
    mapping={
        0: 0,
        1: 0,
        2: 1,
        3: 1,
        4: 2,  # null atom
        5: 2,  # null atom
        6: 2,  # null atom
    },
    name="G",
)
```

```python
derivative = P.derivative(mu, G, return_type="non_param")
```

```python
all(
    P(U & V) == P.derivative(mu, G, return_type="non_param").integrate(U @ V)
    for U, V in zip(F, G)
)
```

```python
import numpy as np

rng = np.random.default_rng(42)

prob_space = sa.ProbabilitySpace.from_rand(
    domain_size=100, num_atoms=25, random_state=rng
)
Omega, F, P = prob_space
G = sa.SigmaAlgebra.from_rand(super=F, num_atoms=15, name="G", random_state=rng)
mu = sa.Measure.from_rand(domain=F, random_state=rng)

P.derivative(mu, G, return_type="non_param")
```

```python
all(
    np.allclose(P(U & V), P.derivative(mu, G, return_type="non_param").integrate(U @ V))
    for U, V in zip(F, G)
)
```

```python
P.derivative(mu, G, return_type="non_param").is_measurable(F @ G)
```

```python
P.derivative(mu, G, return_type="non_param").is_measurable(G @ F)
```
