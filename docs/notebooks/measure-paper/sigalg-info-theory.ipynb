```python
import sigalg as sa

Omega = sa.SampleSpace.from_sequence(size=7)
F = sa.SigmaAlgebra(
    domain=Omega,
    mapping={
        0: 0,
        1: 1,
        2: 2,
        3: 2,
        4: 3,  # null atom
        5: 4,  # null atom
        6: 4,  # null atom
    },
)
P = sa.ProbabilityMeasure(
    domain=F,
    mapping={
        0: 0.2,
        1: 0.6,
        2: 0.2,
        3: 0.0,
        4: 0.0,
    },
)
mu = sa.Measure(
    domain=F,
    mapping={
        0: 1,
        1: 2,
        2: 3,
        3: 4,
        4: 5,
    },
)
G = sa.SigmaAlgebra(
    domain=Omega,
    mapping={
        0: 0,
        1: 0,
        2: 1,
        3: 1,
        4: 2,  # null atom
        5: 2,  # null atom
        6: 2,  # null atom
    },
    name="G",
)
```

```python
print(
    sa.Operators.integrate(
        P.conditional(G).surprisal(mu).integrate(measure=P), measure=P | G
    )
)
```

```python
print(P.conditional(G).surprisal(mu))
```

```python
P.conditional(G).surprisal(mu).integrate(measure=P).to_measurable_vector(
    sig_alg=G, ascend=True
).integrate(measure=P)
```

```python
# P.conditional(G, ascend=True).surprisal(mu).integrate(measure=P)
print(P.conditional(G, ascend=True))
```

```python
X = sa.Domain.from_sequence(size=3)
F = sa.SigmaAlgebra(
    domain=X,
    mapping={
        0: (0, 1),
        1: (1, 1),
        2: (1, 1),
    },
)

Theta = sa.Domain.from_sequence(size=2, variable_name="theta", name="Theta")
g = sa.ParametrizedMeasurableFunction.from_domains(
    measurable_domain=X,
    parameter_domain=Theta,
    sig_alg=F,
    mapping={
        (0, 0): 2,  # (theta, X) = (0, 0)
        (0, 1): 4,
        (0, 2): 4,
        (1, 0): 1,
        (1, 1): -1,
        (1, 2): -1,
    },
    name="g",
)

print(g)
```
