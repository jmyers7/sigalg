The `Operators.pushforward` method in SigAlg is a method for computing *pushforwards* of probability measures along random vectors. The API reference is [here](https://johnmyers-phd.com/sigalg/api/core/#sigalg.core.Operators.pushforward).

## Mathematical definition

Let $X: \Omega \to \mathbb{R}^d$ be a random vector on a probability space $(\Omega, \mathcal{F},P)$. Then we define a probability measure $P_X$ on $\mathbb{R}^d$, called the *pushforward* (or *image*) *measure* of $P$ by setting

$$
P_X(A) = P\left( \{\omega \in \Omega : X(\omega) \in A\}\right),
$$

for all Borel measurable subsets $A\subset \mathbb{R}^d$.

## API examples


We begin by defining a sample space $\Omega = \{0,1,2,3\}$ and a probability measure $P$ on $\Omega$.

```python
from sigalg.core import ProbabilityMeasure, SampleSpace

Omega = SampleSpace().from_sequence(size=4)

P = ProbabilityMeasure(sample_space=Omega).from_dict(
    {
        0: 0.15,
        1: 0.35,
        2: 0.1,
        3: 0.4,
    }
)
```

Now, define a random vector $X$ on $\Omega$, and compute the pushforward:

```python
from sigalg.core import Operators, RandomVector

X = RandomVector(domain=Omega).from_dict(
    {
        0: (1, 2),
        1: (1, 2),
        2: (3, -1),
        3: (0, 1),
    }
)

Operators.pushforward(rv=X, probability_measure=P)
```
