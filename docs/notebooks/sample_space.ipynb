The `SampleSpace` class in SigAlg is the fundamental class for representing sample spaces. The API reference is [here](https://johnmyers-phd.com/sigalg/api/modules/core/#sigalg.core.SampleSpace).

## Mathematical definition

In the abstract, a *sample space* is just a set $\Omega$. However, in the context of probability theory, sample spaces are often conceptualized as the set of all possible outcomes of a random experiment. Each element $\omega \in \Omega$ is called a *sample point* or *outcome*. The sample space serves as the foundational building block for defining events (subsets of $\Omega$) and probability measures (functions that assign probabilities to events).

Sample spaces are not meant to contain data. Instead, data is meant to be encoded in random variables and vectors defined on the sample space, which are functions on sample spaces.

## API examples

### Creating sample spaces

#### From lists

Create a sample space from a list of sample points:

```python
from sigalg.core import SampleSpace

Omega = SampleSpace(name="Omega").from_list(["red", "green", "blue"])
print(Omega)
```

#### From `pd.Index` objects

Create a sample space from a `pd.Index` object:

```python
import pandas as pd

data = pd.Index(["a", "b", "c"], name="sample")
Omega_2 = SampleSpace(name="Omega_2").from_pandas(data=data)
print(Omega_2)
```

#### From sequences

The `from_sequence` method creates a sample space with integer indices:

```python
Omega = SampleSpace().from_sequence(size=5)
print(Omega)
```

You can also specify an initial index and custom prefix for the sample points:

```python
Omega = SampleSpace(name="Omega").from_sequence(size=4, initial_index=3, prefix="omega")
print(Omega)
```

### Accessing sample points

#### Indexing

Access individual sample points by position:

```python
Omega = SampleSpace().from_sequence(size=5)

print(f"First sample point: {Omega[0]}")
print(f"Last sample point: {Omega[-1]}")
```

#### Slicing

Use slicing to extract subsets of the sample space:

```python
Omega = SampleSpace().from_list(["a", "b", "c", "d", "e"])

event = Omega[1:4]
print(event)
```

#### Iteration

Iterate through all sample points:

```python
Omega = SampleSpace().from_sequence(size=3)

for omega in Omega:
    print(f"Sample point: {omega}")
```

### Creating events

Sample spaces can create events, which are subsets of the sample space:

```python
Omega = SampleSpace().from_sequence(size=5)
F = SigmaAlgebra.power_set(Omega)

A = F.get_event([0, 2, 4], name="A")
print(A)
```

### Conversion methods

#### Converting to a probability space

Create a probability space with a uniform distribution:

```python
from sigalg.core import ProbabilityMeasure, SigmaAlgebra

Omega = SampleSpace().from_list(["a", "b", "c"])

# Create with default uniform measure and power-set sigma-algebra
prob_space = Omega.make_probability_space()
print(prob_space)
```

Create a probability space with custom probability measure and $\sigma$-algebra:

```python
# Custom probability measure
probs = {"a": 0.5, "b": 0.3, "c": 0.2}
P = ProbabilityMeasure(sample_space=Omega).from_dict(probs)

# Custom sigma-algebra
F = SigmaAlgebra(sample_space=Omega).from_dict({"a": 0, "b": 1, "c": 1})

# Create probability space
prob_space = Omega.make_probability_space(sig_alg=F, prob_measure=P)
print(prob_space)
```

#### Converting to an event space

Create an event space with default power set $\sigma$-algebra:

```python
Omega = SampleSpace().from_list(["s0", "s1", "s2", "s3"])

event_space = Omega.make_event_space()
print(event_space)
```

Create an event space with a custom $\sigma$-algebra:

```python
# Custom sigma-algebra with two atoms
F = SigmaAlgebra(sample_space=Omega).from_dict({"s0": 0, "s1": 0, "s2": 1, "s3": 1})

event_space = Omega.make_event_space(sig_alg=F)
print(event_space)
```

### Properties of sample spaces

#### Length

Get the number of sample points:

```python
Omega = SampleSpace().from_sequence(size=10)
print(f"Number of sample points: {len(Omega)}")
```

#### Checking membership

Check if a sample point is in the sample space:

```python
Omega = SampleSpace().from_list(["a", "b", "c"])

print(f"Is 'a' in Omega? {'a' in Omega}")
print(f"Is 'd' in Omega? {'d' in Omega}")
```
